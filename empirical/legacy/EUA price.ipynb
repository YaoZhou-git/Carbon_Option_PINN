```
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

# 读取Excel文件
import os
from pathlib import Path
file_path = Path(os.environ['CARBON_OPTION_DATA_DIR']) / 'data202305-202404.xlsx'
df = pd.read_excel(file_path, sheet_name='Sheet1')

# 确保日期列是datetime类型
df['data'] = pd.to_datetime(df['data'])
plt.rcParams.update({
    'font.size': 22,  
    'font.weight': 'bold', 
    'axes.titlesize': 22, 
    'axes.labelsize': 22,
    'axes.titleweight': 'bold', 
    'axes.labelweight': 'bold',  
    'legend.fontsize': 22,

    # 添加全局线宽设置
    'axes.linewidth': 2,         # 坐标轴线宽
    'grid.linewidth': 1.5,      # 网格线宽
    'lines.linewidth': 2.5,     # 曲线线宽
    'xtick.major.width': 2,     # X轴主刻度线宽
    'ytick.major.width': 2,     # Y轴主刻度线宽
    'hatch.linewidth': 2,       # 填充图案线宽
})
```

```

# 创建图形
plt.figure(figsize=(15, 8))
plt.plot(df['data'], df['price'], linewidth=1)

# 设置横坐标格式 - 更精确地控制季度标签
ax = plt.gca()

# 创建季度开始日期的列表
start_date = df['data'].min()
end_date = df['data'].max()
quarters = pd.date_range(start=start_date, end=end_date, freq='3MS')  # 每季度开始

# 设置主要刻度位置
ax.set_xticks(quarters)

# 格式化标签
ax.set_xticklabels([date.strftime('%Y/%m') for date in quarters], rotation=45)

# 添加网格
# plt.grid(True, alpha=0.3)

# 添加标题和标签
# plt.title('Futures Settlement Price on EUA from 2021 to 2023', fontsize=24)
plt.xlabel('Date', fontsize=24)
plt.ylabel('Value', fontsize=24)

# 调整布局
plt.tight_layout()
plt.savefig("Futures Settlement Price on EUA from 2021 to 2023.png", dpi=300, bbox_inches='tight')
# 显示图形
plt.show()
```
